```python
fix_ids = updates[updates["pH"].notna()]["seq_id"]
```

```python
train[train["seq_id"].isin(fix_ids)]
```

```python
train_clean = train.copy()
```

```python
# identifying the na rows. 
delete_ids = updates[updates["pH"].isna()]["seq_id"]
```

```python
train_clean = train_clean[~train_clean["seq_id"].isin(delete_ids)]
```

```python
train_clean.shape
```

```python
31390-2409
```

```python
fixes = updates[updates["pH"].notna()].set_index("seq_id")
```

```python
train_clean = train_clean.set_index("seq_id")
```

```python
#overwriting the values with the correct ones 
train_clean.loc[fixes.index,["pH","tm"]] = fixes[["pH","tm"]]
```

```python
train_clean = train_clean.reset_index()
```

```python
train_clean[train_clean["seq_id"].isin(fix_ids)]
```

After examining the update file, I found 2,409 rows with missing values in every column except seq_id. Following Kaggle's instructions, I deleted these rows from the training data. The other 25 rows correct entries where pH and tm were swapped in the original data, so I overwrote those rows with the corrected values from the update file. The cleaned dataset has 28,981 rows.
