After applying the updates, 286 rows were missing pH. I dropped these rows because protein stability depends on pH, so the same protein can have a different melting temperature at a different pH. Without a recorded pH, a row's tm can't be compared fairly with the others. Since only 286 rows (about 1%) were affected, dropping them keeps the data consistent at little cost.

I also dropped the 968 remaining rows missing a citation in data_source. A missing citation doesn't make a measurement wrong, but it means the value can't be traced back to its source, which makes it less reliable. This concern is supported by the Kaggle update: 2,367 of the 2,409 rows Kaggle officially deleted were uncited. The tm values of the uncited rows were broadly similar to the cited ones, so this was a precaution rather than a response to clearly wrong values. The cleaned dataset now has 27,727 rows.

## Step 6 : Duplicate sequences

```python
train_clean["protein_sequence"].duplicated(keep=False).sum()
```

```python
dup_mask = train_clean.duplicated(subset=["protein_sequence", "pH"], keep=False)
dup_mask.sum()
```

```python
train_clean[dup_mask].sort_values(["protein_sequence", "pH"])[["seq_id", "pH", "tm"]].head(10)
```

```python
dup_groups = train_clean[dup_mask].groupby(["protein_sequence", "pH"])["tm"].agg(["count", "min", "max"])
dup_groups["tm_range"] = dup_groups["max"] - dup_groups["min"]
len(dup_groups)
```

```python
dup_groups["tm_range"].describe()
```

```python
(dup_groups["tm_range"] == 0).sum()
```

740 duplicate rows form 341 groups (so about 2 copies per group, on average) and only 43 groups agree exactly.There is an explination for this we can see that half the groups disagree by 1.7°C or less. That's normal measurement noise between labs or repeat runs.The problem is that some disagree way too much for example seq_ids 2239 and 2240: same sequence, same pH 4.5, but tm 83.6 versus 34.1. That's a 49°C gap, and the max is 58.3°C. This is a problem both can't be right, and we can't tell which one is correct the best thing to do here is drop the pair. 

```python
(dup_groups["tm_range"]>5).sum()
```

Some proteins were measured more than once at the same pH. Leaving these copies in causes two problems. First, those proteins count twice in any summary or plot. Second, the data contradicts itself: for example, seq_ids 1764 and 1765 have the same sequence and pH but tm values of 53.1 and 47.1°C.

To fix this, I combined each set of copies into a single row and used the median tm as the best estimate. When copies disagreed by more than 5°C, I dropped them instead, because the measurements conflict and there is no way to tell which is correct. I chose 5°C because it is above typical measurement noise but below the clearly conflicting cases, such as seq_ids 2239 and 2240, which differ by 49°C.

```python
# Combine rows with the same sequence and pH into one row.
# Rows that appear only once pass through unchanged.
# tm_min and tm_max are kept temporarily to measure disagreement.

train_dedup = train_clean.groupby(["protein_sequence", "pH"], as_index=False).agg(
    seq_id=("seq_id", "first"),
    data_source=("data_source", "first"),
    tm=("tm", "median"),
    tm_min=("tm", "min"),
    tm_max=("tm", "max"),
)
train_dedup.shape
```

```python
# Drop groups whose copies disagree by more than 5 degrees C, then remove helper columns.
train_dedup = train_dedup[(train_dedup["tm_max"] - train_dedup["tm_min"]) <= 5]
train_dedup = train_dedup.drop(columns=["tm_min", "tm_max"])
train_dedup.shape
```

In summary of the 740 duplicate rows, 43 groups agreed exactly and 106 disagreed by more than 5°C and were dropped. The remaining groups were combined using their median tm. The final cleaned dataset has 27,222 rows.
